```python
df.dropna(inplace=True)
print("Missing values after cleaning:")
print(df.isna().sum())
```

```python
print("Duplicate values in Dataset:",df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
print("Duplicate values in Dataset after Removing duplicates:",df.duplicated().sum())
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
numeric_columns = ["Open", "High", "Low", "Close", "Adj Close", "Volume"]
for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```

```python
print("Sort Data by Date")
df.sort_values("Date", inplace=True)
df.head()
```

```python
before = len(df)
df = df[(df["Open"] > 0)&(df["High"] > 0)&(df["Low"] > 0)&(df["Close"] > 0)&(df["Volume"] >= 0)].copy()
df.reset_index(drop=True, inplace=True)
after = len(df)
print("Invalid values removed:", before - after)
print("Remaining records:", after)
```

```python
print("Available Stocks:", df["Ticker"].unique())
ticker = input("Enter Ticker(AAPL/MSFT/NFLX): ").upper()
if ticker in df["Ticker"].unique():
    df = df[df["Ticker"] == ticker].copy()
    print("Selected Stock:", ticker)
    print("Records:", len(df))
else:
    print("Invalid ticker.")
```

```python
df.sort_values("Date", inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
date = input("Enter date (YYYY-MM-DD): ")
date = pd.to_datetime(date)
result = df[df["Date"] == date]
if result.empty:
    print("No data found for this date.")
else:
    print(result)
```

```python
start_date = pd.to_datetime(input("Enter start date (YYYY-MM-DD): "))
end_date = pd.to_datetime(input("Enter end date (YYYY-MM-DD): "))
result = df[(df["Date"] >= start_date)&(df["Date"] <= end_date)]
if result.empty:
    print("No data found for this date range.")
else:
    print(result)
```

```python
df["Daily_Return"] = df["Close"].pct_change() * 100
print("Average Return:", df["Daily_Return"].mean())
print("Highest Return:", df["Daily_Return"].max())
print("Lowest Return:", df["Daily_Return"].min())
print("Return Standard Deviation:", df["Daily_Return"].std())
positive_days = (df["Daily_Return"] > 0).sum()
negative_days = (df["Daily_Return"] < 0).sum()
print("Positive Days:", positive_days)
print("Negative Days:", negative_days)
```
