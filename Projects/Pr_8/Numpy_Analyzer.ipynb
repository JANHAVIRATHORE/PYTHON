```python
import numpy as np
```

```python
class DataAnalyzer:

    def __init__(self):
        self.array=np.array([])

    def create_1d_array(self):  
        lst=[]
        length = int(input("Enter the size of the 1D array:"))
        for i in range(length):
            element = int(input(f"Enter {i} element {i+1}:"))
            lst.append(element)
        self.array=np.array(lst)
        print(f"Created 1D array: {self.array}")

    def create_2d_array(self):
        rows = int(input("Enter the number of rows:"))
        cols = int(input("Enter the number of columns:"))
        lst=[]
        for i in range(rows):
            row=[]
            for j in range(cols):
                element = int(input(f"Enter ({i},{j}) element at position: "))
                row.append(element)
            lst.append(row)
        self.array=np.array(lst)
        print(f"Array Created Successfully:\n {self.array}")

    def create_3d_array(self):
        depth = int(input("Enter the depth of the 3D array:"))
        rows = int(input("Enter the number of rows:"))
        cols = int(input("Enter the number of columns:"))
        lst=[]
        for i in range(depth):
            layer=[]
            for j in range(rows):
                row=[]
                for k in range(cols):
                    element = int(input(f"Enter ({i},{j},{k}) element at position: "))
                    row.append(element)
                layer.append(row)
            lst.append(layer)
        self.array=np.array(lst)
        print(f"Array Created Successfully:\n {self.array}")

    def __check_array(self):
        if self.array.size == 0:
            print("Please create an array first.")
            return False
        return True

    def indexing(self):
        if not self.__check_array():
            return
        print("\nCurrent Array:")
        print(self.array)

        print("\nExamples:")
        print("1D array → 2")
        print("2D array → 1,2")
        print("3D array → 0,1,2")
        index_input = input("\nEnter index separated by commas: ")
        try:
            indexes = tuple(map(int, index_input.split(",")))
            if len(indexes) != self.array.ndim:
                print(f"{self.array.ndim}D array.\nEnter {self.array.ndim} index(es).")
                return
            result = self.array[indexes]
            print("Indexed value:", result)
        except (IndexError, ValueError):
            print("Invalid index.")

    def slicing(self, start, end):
        if not self.__check_array():
            return
        print("\nCurrent Array:")
        print(self.array)
        print("Sliced array:", self.array[start:end])

    def add(self):
        if not self.__check_array():
            return
    
        num=int(input("Enter number to add: "))
        result=self.array+num
        print("Original array:",self.array)
        print("After addition:",result)  


    def sub(self):
        if not self.__check_array():
            return
        num=int(input("Enter number to subtract: "))
        result=self.array-num
        print("Original array:",self.array)
        print("After subtraction:",result)  

    def multi(self):
        if not self.__check_array():
            return
        num=int(input("Enter number to multiply: "))
        result=self.array*num
        print("Original array:",self.array)
        print("After multiplication:",result)    

    def div(self):
        if not self.__check_array():
            return
        num=int(input("Enter number to divide by: "))
        if num==0:
            print("Division by zero is not allowed.")
            return
        result=self.array/num
        print("Original array:",self.array)
        print("After division:",result)

    def combine(self):
        if not self.__check_array():
            return
        print("Current array:")
        print(self.array)
        second=input("Enter another array elements separated by spaces:")
        try:
            second_array=np.array(list(map(int, second.split())))
            result = np.concatenate((self.array.flatten(),second_array))
            print("Combined array:", result)
        except ValueError:
            print("Please enter only numbers.")
            
    def split(self):
        if not self.__check_array():
            return
        part=int(input("Enter number of parts: "))
        try:
            result=np.array_split(self.array,part)
            print("Split arrays:")
            for i, part in enumerate(result):
                print(f"Part {i+1}: {part}")
        except ValueError:
            print("Unable to split the array.")

    def search(self):
        if not self.__check_array():
            return
        num=int(input("Enter the number to search:"))
        result=np.where(self.array==num)
        print("Value index:", result)

    def filter_array(self):
        if not self.__check_array():
            return
        num=int(input("Enter the number to filter:"))
        filter_arr=self.array[self.array>num]
        print("Filtered array greater than", num, ":", filter_arr)
        filter_arr=self.array[self.array<num]
        print("Filtered array less than", num, ":", filter_arr)

    def sort_array(self):
        if not self.__check_array():
            return
        print("Original array:", self.array)
        sort_arr=np.sort(self.array)
        print("Sorted array:", sort_arr)  
        reverse_arr = np.sort(self.array)[::-1]
        print("Reverse sorted array:", reverse_arr)                               

print("\nWelcome to the Numpy Analyzer!")
print(f"="*50,"\n")
obj=DataAnalyzer()

while True:
    print("\nChoose an option:\n")
    print("1.Create a NumPy array")
    print("2.Indexing and slicing")
    print("3.Perform Mathematical operations")
    print("4.Combine and Split Arrays")
    print("5.Search, Sort, or Filter Arrays")
    print("6.Compute Aggregates and Statisticas")
    print("7.Exit")

    choice=int(input("\nEnter your choice"))

    if choice==1:
      while True:
        print("\nSelect the type of arrray to create:\n")
        print("1.1D array")
        print("2.2D array")
        print("3.3D array")
        print("4.Back to main menu")

        choice=int(input("\nEnter your choice"))

        if choice==1:
            obj.create_1d_array()       
            
        elif choice==2:
            obj.create_2d_array()
            
        elif choice==3:
            obj.create_3d_array()
            
        elif choice==4:
            break

        else:
            print("\nInvalid choice. Please try again.")

    elif choice==2:
        while True:
            print("\nChoose an operation:\n")
            print("1.Indexing")
            print("2.Slicing")
            print("3.Go Back")
        
            choice=int(input("\nEnter your choice"))
        
            if choice==1:
                obj.indexing()       
                    
            elif choice==2:
                start=int(input("\nEnter starting index:"))
                end=int(input("Enter end index:"))
                obj.slicing(start,end)

            elif choice==3:
                break
        
            else:
               print("\nInvalid choice. Please try again.")
        
    elif choice==3:
        while True:
            print("\nChoose a mathematical operation:\n")
            print("1.Addition")
            print("2.Subtraction")
            print("3.Multiplication")
            print("4.Division")
            print("5.Go Back")
        
            choice=int(input("\nEnter your choice"))
        
            if choice==1:
                obj.add()
        
            elif choice==2:
                obj.sub()
                    
            elif choice==3:
                obj.multi()

            elif choice==4:
                obj.div()

            elif choice==5:
                break

            else:
                print("\nInvalid choice. Please try again.")

    elif choice==4:
        while True:
            print("\nChoose a option:\n")
            print("1.Combine Arrays")
            print("2.Split Array")
            print("3.Go Back")
                
            choice=int(input("\nEnter your choice"))
                
            if choice==1:
                obj.combine()
                
            elif choice==2:
                obj.split()
                            
            elif choice==3:
                break
        
            else:
                print("\nInvalid choice. Please try again.")

    elif choice==5:
        while True:
            print("\nChoose a option:\n")
            print("1.Search a Value")
            print("2.Sort the Array")
            print("3.Filter Values")
            print("4.Go Back")
                
            choice=int(input("\nEnter your choice"))
                
            if choice==1:
                obj.search()
                
            elif choice==2:
                obj.sort_array()
                            
            elif choice==3:
                obj.filter_array()

            elif choice==4:
                break
        
            else:
                print("\nInvalid choice. Please try again.")
        

    elif choice==6:
        while True:
            print("\nChoose a aggregate/statistical operation:\n")
            print("1.Sum")
            print("2.Mean")
            print("3.Median")
            print("4.Standard Deviation")
            print("5.Variance")
            print("6.Go Back")
                
            choice=int(input("\nEnter your choice"))
                
            if choice==1:
                if not obj._DataAnalyzer__check_array():
                    continue
                total= np.sum(obj.array)
                print("\nSum of Array:",total)

            elif choice==2:
                if not obj._DataAnalyzer__check_array():
                    continue
                mean=np.mean(obj.array)
                print("\nMean of Array:",mean)

            elif choice==3:
                if not obj._DataAnalyzer__check_array():
                    continue
                median=np.median(obj.array)
                print("\nMedian of Array:",median)
        
            elif choice==4:
                if not obj._DataAnalyzer__check_array():
                    continue
                std_dev=np.std(obj.array)
                print("\nStandard Deviation of Array:",std_dev)
        
            elif choice==5:
                if not obj._DataAnalyzer__check_array():
                    continue
                variance=np.var(obj.array)    
                print("\nVariance of Array:",variance)

            elif choice==6:
                break
        
            else:
                print("\nInvalid choice. Please try again.")
        

    elif choice==7:
        print("\nThank you for using the Numpy Analyzer! Goodbye!")
        break

    else:
        print("\nInvalid choice. Please try again.")
```
