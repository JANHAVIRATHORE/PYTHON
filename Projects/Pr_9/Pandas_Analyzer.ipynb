```python
import pandas as pd
import matplotlib.pylab as plt
import seaborn as sns
import numpy as np
```

```python

class SalesDataAnalyzer:

    def __init__(self):
        self.data = None
        self.df = None
        self.df2 = None
        self.fig = None
    def load_dataset(self, data):
        try:
            self.data = data
            self.df = pd.read_csv(self.data)
            print("\nDataset loaded successfully!")
            print(f"Rows: {self.df.shape[0]}, Columns: {self.df.shape[1]}")
        except FileNotFoundError:
            print("\nFile not found.")
            self.data = None
            self.df = None

    def head_row(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.head())   

    def last_row(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.tail()) 

    def col_name(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.columns)        

    def data_type(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.dtypes)    

    def basic_info(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        self.df.info()   

    def add_cols(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        colname=input("Enter new column name:")
        value=input("Enter value for the new column: ")
        self.df[colname]=value
        print("\nNew column added successfully!")
        print(self.df.head())

    def drop_cols(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        col_name = input("Enter column name to drop: ")
        if col_name in self.df.columns:
            self.df.drop(columns=col_name, inplace=True)
            print("\nColumn dropped successfully!")
        else:
            print("\nColumn not found!")

    def rename_cols(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        old_name=input("Enter existing column name:")
        new_name=input("Enter new column name:")
        if old_name in self.df.columns:
            self.df.rename(columns={old_name:new_name}, inplace=True)
            print("\nColumn renamed successfully!")
        else:
            print("\nColumn not found!")

    def group_data(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.columns)
        group_column=input("Enter group column: ")
        value_column=input("Enter numeric value column: ")
        if group_column not in self.df.columns:
            print("\nGroup column does not exist!")
            return
        if value_column not in self.df.columns:
            print("\nValue column does not exist!")
            return
        if self.df[value_column].dtype not in ["int64", "float64"]:
            print("\nPlease select a numeric value column!")
            return
        print("\nSum:")
        print(self.df.groupby(group_column)[value_column].sum())
        print("\nMean:")
        print(self.df.groupby(group_column)[value_column].mean())
        print("\nCount:")
        print(self.df.groupby(group_column)[value_column].count())
        print("\nMinimum:")
        print(self.df.groupby(group_column)[value_column].min())
        print("\nMaximum:")
        print(self.df.groupby(group_column)[value_column].max())
        self.df["Group_Mean"] = self.df.groupby(group_column)[value_column].transform("mean")
        print("\nGroup Mean using transform:")
        print(self.df[[group_column, value_column, "Group_Mean"]])

    def pivottable(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.columns)
        index_column=input("Enter index column: ")
        value_column=input("Enter value column: ")
        if index_column not in self.df.columns:
            print("\nIndex column does not exist!")
            return
        if value_column not in self.df.columns:
            print("\nValue column does not exist!")
            return
        print("\nSum:")
        print(self.df.pivot_table(index=index_column,values=value_column,aggfunc="sum"))
        print("\nMean:")
        print(self.df.pivot_table(index=index_column,values=value_column,aggfunc="mean"))
        print("\nCount:")
        print(self.df.pivot_table(index=index_column,values=value_column,aggfunc="count"))
        print("\nMinimum:")
        print(self.df.pivot_table(index=index_column,values=value_column,aggfunc="min"))
        print("\nMaximum:")
        print(self.df.pivot_table(index=index_column,values=value_column,aggfunc="max"))

    def re_index(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print("\nCurrent index:")
        print(self.df.index.tolist())
        values=input("Enter new index values separated by comma: ")
        try:
            new_index=[int(x.strip()) for x in values.split(",")]
            self.df=self.df.reindex(new_index)
            print("\nDataFrame reindexed successfully!")
            print(self.df)
        except ValueError:
            print("\nPlease enter valid index numbers!")

    def show_data(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print("\nCurrent DataFrame:")
        print(self.df)

    def indexing(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        numeric_data = self.df.select_dtypes(include="number").to_numpy()
        print("\nNumPy Array:")
        print(numeric_data)
        row=int(input("Enter row index: "))
        col=int(input("Enter column index: "))
        print("\nSelected Value:")
        print(numeric_data[row, col])

    def slicing(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        numeric_data = self.df.select_dtypes(include="number").to_numpy()
        print("\nNumPy Array:")
        print(numeric_data)
        rows=int(input("Enter number of rows to slice: "))
        cols=int(input("Enter number of columns to slice: "))
        print("\nSliced NumPy Array:")
        print(numeric_data[:rows, :cols])
            
    def mathematical_operations(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print("\nSum:")
        print(self.df.sum(numeric_only=True))
        value=float(input("\nEnter value for mathematical operations: "))
        try:
            print("\nSubtraction:")
            print(self.df.select_dtypes(include="number") - value)
            print("\nMultiplication:")
            print(self.df.select_dtypes(include="number") * value)
            print("\nDivision:")
            print(self.df.select_dtypes(include="number") / value)
        except ZeroDivisionError:
            print("\nInvalid division")            

    def Concate_data(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        result = pd.concat([self.df, self.df2],ignore_index=True)
        print("\nConcatenated DataFrame:")
        print(result)

    def merge_data(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        column_name = input("Enter common column name: ")
        if column_name in self.df.columns and column_name in self.df2.columns:
            result = pd.merge(self.df, self.df2,on=column_name)
            print("\nMerged DataFrame:")
            print(result)
        else:
            print("\nCommon column not found!")    

    def join_data(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        column_name = input("Enter common column name: ")
        if column_name in self.df.columns and column_name in self.df2.columns:       
            result = self.df.join(self.df2,lsuffix="_left",rsuffix="_right")
            print("\nJoined DataFrame:")
            print(result)

    def split_dataframe(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.columns)
        column_name=input("Enter column to split by: ")
        if column_name not in self.df.columns:
            print("\nColumn not found!")
            return
        value=input("Enter value: ")
        df1=self.df[self.df[column_name].astype(str) == value]
        df2=self.df[self.df[column_name].astype(str) != value]
        print("\nFirst DataFrame:")
        print(df1)
        print("\nSecond DataFrame:")
        print(df2)
        
    def filter_value(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.columns)
        column_name=input("Enter column name: ")
        if column_name not in self.df.columns:
            print("\nColumn does not exist!")
            return
        if self.df[column_name].dtype in ["int64", "float64"]:
            value=float(input("Enter value: "))
            result=self.df[self.df[column_name] > value]
        else:
            value=input("Enter text value: ")
            result=self.df[self.df[column_name].astype(str).str.lower()==value.lower()]
        print("\nFiltered Data:")
        print(result)

    def search_value(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.columns)
        column_name=input("Enter column name: ")
        if column_name not in self.df.columns:
            print("\nColumn does not exist!")
            return
        search_value=input("Enter search text: ")
        result=self.df[self.df[column_name].astype(str).str.lower()==search_value.lower()]
        print("\nSearch Result:")
        print(result)

    def sort_dataframe(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.columns)
        column_name = input("Enter column name to sort: ")
        if column_name not in self.df.columns:
            print("\nColumn not found!")
            return
        order = input("Enter order (ascending as a/descending as d): ")
        if order.lower() == "a":
            self.df = self.df.sort_values(by=column_name, ascending=True)
            print("\nData sorted in ascending order!")
        elif order.lower() == "d":
            self.df = self.df.sort_values(by=column_name, ascending=False)
            print("\nData sorted in descending order!")
        else:
            print("\nInvalid sorting order!")
            return
        print(self.df)
        
    def aggregate_function(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print("\nSum:")
        print(self.df.sum(numeric_only=True))
        print("\nMean:")
        print(self.df.mean(numeric_only=True))
        print("\nCount:")
        print(self.df.count(numeric_only=True))

    def statistical_analysis(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print("\nDescriptive Statistics:",self.df.describe())
        print("\nStandard Deviation:",self.df.std(numeric_only=True))
        print("\nVariance:",self.df.var(numeric_only=True))
        print("\nPercentile:")
        print("25%:", self.df.quantile(0.25, numeric_only=True))
        print("50%:", self.df.quantile(0.50, numeric_only=True))
        print("75%:", self.df.quantile(0.75, numeric_only=True))
        print("\nQuantile:")
        print("Q1:", self.df.quantile(0.25, numeric_only=True))
        print("Q2:", self.df.quantile(0.50, numeric_only=True))
        print("Q3:", self.df.quantile(0.75, numeric_only=True))

    def missing_value(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print(self.df.isna())   
        print("\nTotal missing values per column:\n",self.df.isna().sum())

    def fill_value(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        self.df.fillna(self.df.mean(numeric_only=True),inplace=True)   
        print("\nMissing values filled with mean successfully!")

    def drop_missing(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        self.df.dropna(inplace=True)
        print("\nRows with missing values dropped successfully!")

    def replace_missing(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        value=input("Enter the value to fill missing values:")
        self.df.fillna(value,inplace=True)
        print("\nMissing values replaced successfully!")

    def plot_bar(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        columns=self.df.select_dtypes(include="number").columns.tolist()
        print("\nNumeric columns:", columns)
        column_name=input("Enter numeric column for Bar Plot: ")
        if column_name not in columns:
            print("\nColumn not found!")
            return
        sns.set_style("whitegrid")
        self.fig=plt.figure(figsize=(10, 5))
        sns.barplot(x=self.df.index,y=self.df[column_name])
        plt.xlabel("Index")
        plt.ylabel(column_name)
        plt.title(f"Bar Plot - {column_name}")
        plt.tight_layout()
        plt.show()

    def plot_line(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        columns=self.df.select_dtypes(include="number").columns.tolist()
        print("\nNumeric columns:", columns)
        column_name=input("Enter numeric column for Line Plot: ")
        if column_name not in columns:
            print("\nColumn not found!")
            return
        self.fig=plt.figure(figsize=(10, 5))
        plt.plot(self.df.index, self.df[column_name])
        plt.xlabel("Index")
        plt.ylabel(column_name)
        plt.title(f"Line Plot - {column_name}")
        plt.tight_layout()
        plt.show()

    def plot_scatter(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        columns=self.df.select_dtypes(include="number").columns.tolist()
        print("\nNumeric columns:", columns)
        x_column=input("Enter X-axis column: ")
        y_column=input("Enter Y-axis column: ")
        if x_column not in columns or y_column not in columns:
            print("\nColumn not found!")
            return
        self.fig=plt.figure(figsize=(8, 5))
        plt.scatter(self.df[x_column], self.df[y_column])
        plt.xlabel(x_column)
        plt.ylabel(y_column)
        plt.title(f"Scatter Plot - {x_column} vs {y_column}")
        plt.tight_layout()
        plt.show()

    def plot_pie(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        print("\nAvailable columns:", self.df.columns.tolist())
        column_name=input("Enter column name: ")
        if column_name not in self.df.columns:
            print("\nColumn does not exist!")
            return
        data=self.df[column_name].value_counts()
        self.fig=plt.figure(figsize=(8, 5))
        plt.pie(data, labels=data.index, autopct="%1.1f%%")
        plt.title("Pie Chart")
        plt.show()

    def plot_histogram(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        columns=self.df.select_dtypes(include="number").columns.tolist()
        print("\nNumeric columns:", columns)
        column_name=input("Enter numeric column for Histogram: ")
        if column_name not in columns:
            print("\nColumn not found!")
            return
        self.fig=plt.figure(figsize=(8, 5))
        plt.hist(self.df[column_name].dropna(), bins=10)
        plt.xlabel(column_name)
        plt.ylabel("Frequency")
        plt.title(f"Histogram - {column_name}")
        plt.tight_layout()
        plt.show()

    def plot_stack(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        columns=self.df.select_dtypes(include="number").columns.tolist()
        print("\nNumeric columns:", columns)
        column1=input("Enter first numeric column: ")
        column2=input("Enter second numeric column: ")
        if column1 not in columns or column2 not in columns:
            print("\nPlease select numeric columns!")
            return
        x=range(len(self.df))
        self.fig=plt.figure(figsize=(8, 5))
        plt.stackplot(x,self.df[column1],self.df[column2],labels=[column1, column2])
        plt.xlabel("Index")
        plt.ylabel("Value")
        plt.title("Stack Plot")
        plt.legend()
        plt.show()

    def plot_box(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        columns=self.df.select_dtypes(include="number").columns.tolist()
        print("\nNumeric columns:", columns)
        column_name=input("Enter numeric column for Box Plot: ")
        if column_name not in columns:
            print("\nColumn not found!")
            return
        self.fig=plt.figure(figsize=(8, 5))
        sns.boxplot(y=self.df[column_name])
        plt.ylabel(column_name)
        plt.title(f"Box Plot - {column_name}")
        plt.tight_layout()
        plt.show()

    def plot_heatmap(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        columns=self.df.select_dtypes(include="number").columns.tolist()
        if len(columns) == 0:
            print("\nNo numeric columns available!")
            return
        self.fig=plt.figure(figsize=(8, 5))
        correlation = self.df[columns].corr()
        sns.heatmap(correlation, annot=True)
        plt.title("Correlation Heatmap")
        plt.tight_layout()
        plt.show()

    def save_visualization(self):
        if self.df is None:
            print("\nPlease load the dataset first!")
            return
        if self.fig is None:
            print("\nPlease generate a visualization first!")
            return
        file_name=input("Enter file name: ")
        self.fig.savefig(file_name + ".png")
        print("\nVisualization saved successfully!")

    def __del__(self):
        self.data=None
        self.df=None
        self.df2=None
        self.fig=None
```

```python


obj=SalesDataAnalyzer()
while True:
    print("\n========== Data Analysis & Visualization Program ==========")
    print("Please select an option")
    print("1.Load Dataset")
    print("2.Explore Data")
    print("3.Perform DataFrame Operations")
    print("4.Handle Missing Data")
    print("5.Generate Descriptive statistics")
    print("6.Data Visualization")   
    print("7.Save Visualization")
    print("8.Exit")    
    print("=============================================================")

    choice=int(input("\nEnter your choice:"))
    if choice==1:
        print("\n==Load Dataset==")
        data=input("Enter the path of the dataset (CSV file,eg:Sales_Data.csv):")
        obj.load_dataset(data)     

    elif choice==2:
     while True:
        print("\n== Explore Data ==")
        print("1.Display the first 5 rows")
        print("2.Display the last 5 rows")
        print("3.Display column names")
        print("4.Display data types")
        print("5.Display basic info")
        print("6.Exit")
        print("=============================================================\n")

        choice= int(input("\nEnter your choice:"))

        if choice==1:
            print("\nFirst 5 rows of dataset is:\n")
            obj.head_row()

        elif choice==2:
            print("\nLast 5 rows of dataset is:\n")
            obj.last_row()        

        elif choice==3:
            print("\nColumn names of dataset is:\n")
            obj.col_name()

        elif choice==4:
            print("\nData Types of dataset is:\n")
            obj.data_type()

        elif choice==5:
            print("\nBasic Info of dataset is:\n")
            obj.basic_info()
        
        elif choice==6:
            break

        else:
            print("\nInvalid choice. Please try again!")

    elif choice==3:
     while True:
        print("\n== Perform DataFrame Operations ==")
        print("1. Creation & DataFrame Operations")
        print("2. Indexing")
        print("3. Slicing")
        print("4. Mathematical Operation (Elementwise)")
        print("5. Combine DataFrame (Concatenate, Merge, Join)")
        print("6. Split")
        print("7. Filter Array, Search")
        print("8. Sort")
        print("9. Aggregate Function (Sum, Mean, Count)")
        print("10. Exit")
        print("=============================================================")

        choice = int(input("\nEnter your choice: "))

        if choice==1:
         while True:
            print("\n== Creation & DataFrame Operations ==")
            print("1. Add a new column")
            print("2. Drop a column")
            print("3. Rename a column")
            print("4. GroupBy")
            print("5. Pivot Table")
            print("6. Reindex")
            print("7. Display Data")
            print("8. Back")
            print("=============================================================")

            choice = int(input("\nEnter your choice: "))   
            if choice==1:
                obj.add_cols()

            elif choice==2:
                obj.drop_cols()
            
            elif choice==3:
                obj.rename_cols()
            
            elif choice==4:
                obj.group_data()
            
            elif choice==5:
                obj.pivottable()
            
            elif choice==6:
                obj.re_index()
            
            elif choice==7:
                obj.show_data()

            elif choice==8:
                break
            
            else:
                print("\nInvalid choice. Please try again!")
            
        elif choice==2:
            obj.indexing()

        elif choice==3:
            obj.slicing()

        elif choice==4:
            obj.mathematical_operations()

        elif choice==5:
         while True:
            print("\n== Combine DataFrame ==")
            print("1. Concatenate")
            print("2. Merge")
            print("3. Join")
            print("4.Exit")
            choice = int(input("Enter choice: "))
            
            if choice==4:
                break
         
            if choice in [1, 2, 3]:    
                file_path = input("Enter second CSV file path: ")
                try:
                    obj.df2 = pd.read_csv(file_path)
                except Exception as e:
                    print(f"\nError loading second dataset: {e}")
                    continue

            if choice==1:
                obj.Concate_data()

            elif choice==2:
                obj.merge_data()
                
            elif choice==3:
                obj.join_data()

            else:
                print("\nInvalid choice!")   

        elif choice==6:
            obj.split_dataframe()

        elif choice==7:
         while True:
            print("1. Filter numeric value")
            print("2. Search text")
            print("3.Exit")
            choice=int(input("\nEnter your choice: "))
            if choice==1:
                obj.filter_value()

            elif choice==2:
                obj.search_value()   

            elif choice==3:
                break
            
            else:
                print("\nInvalid choice. Please try again!")

        elif choice==8:
            obj.sort_dataframe()

        elif choice==9:
            obj.aggregate_function()

        elif choice==10:
            break

        else:
            print("\nInvalid choice. Please try again!")

    elif choice==4:
     while True:
        print("\n== Handle Missing Data ==")
        print("1.Display rows with missing values")
        print("2.Fill missing values with mean")
        print("3.Drop rows with missing values")
        print("4.Replace missing values with a specific value")
        print("5.Exit")
        print("=============================================================\n")

        choice=int(input("\nEnter your choice:"))
        if choice==1:
            print("\nRows with missing values in dataset:\n")
            obj.missing_value()
        
        elif choice==2:
            print("\nFill missing values in dataset:\n")
            obj.fill_value()
        
        elif choice==3:
            print("\nDrop rows with missing values in dataset:\n")
            obj.drop_missing()
        
        elif choice==4:
            print("\nReplace missing values with a specific value in dataset:\n")
            obj.replace_missing()
                
        elif choice==5:
            break

        else:
            print("\nInvalid choice. Please try again!")

    elif choice==5:
        obj.statistical_analysis()

    elif choice==6:
     while True:
        print("\n== Data Visualization ==")
        print("1.Bar plot")
        print("2.Line plot")
        print("3.Scatter plot")
        print("4.Pie plot")
        print("5.Histogram plot")
        print("6.Stack plot")
        print("7.Box plot")
        print("8.Heatmap")
        print("9.Exit")
        print("=============================================================\n")
    
        choice=int(input("\nEnter your choice:"))
        if choice==1:
            obj.plot_bar()
            
        elif choice==2:
            obj.plot_line()        
            
        elif choice==3:
            obj.plot_scatter()     
            
        elif choice==4:
            obj.plot_pie()     
                        
        elif choice==5:
            obj.plot_histogram()     
            
        elif choice==6:
            obj.plot_stack()     

        elif choice==7:
            obj.plot_box()

        elif choice==8:
            obj.plot_heatmap()

        elif choice==9:
            break 

        else:
            print("\nInvalid choice. Please try again!")

    elif choice==7:
        obj.save_visualization()
        
    elif choice==8:
        print("\nExiting the program. Goodbye!")
        break

    else:
        print("\nInvalid choice. Please try again.")
```
